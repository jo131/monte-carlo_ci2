```python
import numpy as np
import math
import matplotlib.pyplot as plt
```

```python
def X(p):
    if np.random.random()<p:
        return 0
    return 1
```

```python
s=0
for i in range(500):
    s += X(0.5)
print(s)
```

```python
def X_geo(lamb):
    al = np.random.random()
    p = np.exp(-lamb)
    k = 0
    while al > p:
        k += 1
        p = p + np.exp(-lamb)*lamb**k/math.factorial(k)
    return k
```

```python
for i in range(500):
    x,y = np.random.uniform(-1, 1),np.random.uniform(-1, 1)
    if x**2 + y**2 <= 1:
        plt.scatter(x,y)
plt.show()
```

```python
def X_inv(lam, u):
    if u<0.5:
        return math.log(2*u)/lam
    return -math.log(2*(1-u))/lam
```
